### 0. Configuración e Importaciones

```python
import pyttsx3, whisper, os
import speech_recognition as sr
from gtts import gTTS
from textblob import TextBlob
from IPython.display import Audio, display

# Initialize the recognizer
recognizer = sr.Recognizer()
print("Setup complete.")
```

### 1. Local Text to Speech (TTS)

```python
def local_tts(text):
    print("Local TTS (Offline)")
    engine = pyttsx3.init()
    # Attempt to set an English voice
    for voice in engine.getProperty('voices'):
        if "ENGLISH" in voice.name.upper():
            engine.setProperty('voice', voice.id)
            break
    engine.say(text)
    engine.runAndWait()
    print(f"Spoken: {text}")

local_tts("Hello, this is your local voice speaking.")
```

### 2. External API Text to Speech (TTS)
Utiliza la API de Google Translate para generar un archivo de audio de alta calidad.

```python
def api_tts(text, filename="api_output.mp3"):
    print("API TTS (Online)")
    tts = gTTS(text=text, lang='en')
    tts.save(filename)
    print(f"Audio saved to: {filename}")
    return filename

file = api_tts("This is the external Google API voice.")
display(Audio(file, autoplay=True))
```

### 3. External API Speech to Text (STT)
Escucha por el micrófono y reconoce el habla usando el servicio de Google.

```python
def api_stt():
    print("API STT (Google)")
    with sr.Microphone() as source:
        print("Listening... Speak now.")
        recognizer.adjust_for_ambient_noise(source, duration=1)
        audio_data = recognizer.listen(source)
    try:
        text = recognizer.recognize_google(audio_data, language="en-US")
        print(f"Google understood: {text}")
        return text, audio_data
    except Exception as e:
        print(f"Error: {e}")
        return None, None

texto_api, audio = api_stt()
```
