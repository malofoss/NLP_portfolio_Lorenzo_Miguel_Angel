### 1. Local Text to Speech (TTS)

```python
def local_tts(text):
    print("Local TTS (Offline)")
    engine = pyttsx3.init()
    # Attempt to set an English voice
    for voice in engine.getProperty('voices'):
        if "ENGLISH" in voice.name.upper():
            engine.setProperty('voice', voice.id)
            break
    engine.say(text)
    engine.runAndWait()
    print(f"Spoken: {text}")

local_tts("Hello, this is your local voice speaking.")
```
